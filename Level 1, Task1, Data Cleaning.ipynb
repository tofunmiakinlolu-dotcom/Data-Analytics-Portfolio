```python
# Data Cleaning and Preprocessing 
In this task, I cleaned the dataset by checking for missing values, removing duplicates, fixing inconsistencies, and changed numerical columns for better understanding and preparing the data for analysis.
```

```python
import pandas as pd
df= pd. read_csv("3) Sentiment dataset.csv")
df.head()
```

```python
df.duplicated().sum()
```

```python
## Handling Duplicates.
I checked for duplicate values, none were found. this indicates that theis dataset is already cleaned in terms of duplication. 
```

```python
df.isnull().sum()
```

```python
## Checking for missing values. 
I checked for missing values in this dataset and i found none. 
```

```python
text_cols = ['Sentiment', 'Platform', 'Country','Text']
df[text_cols] = df[text_cols].apply(lambda x: x.str.strip().str.lower())
df.head()
```

```python
## Fixing Inconsistencies. 
I observed some inconsistencies in some of the categorical columns, such as the Sentiment, Platform, Country, and Text, where values have different letter cases. 
To ensure uniformity and consistency, i converted all entries in these column to lower case letters. 
```

```python
df['Likes'] = df['Likes'].astype(int)
df['Retweets'] = df['Retweets'].astype(int)
df.head()
```

```python
## Data Type Conversion. 
I noticed some numerical columns, such as likes and retweets, which were stored in float formart, so i converted them to integers, to make the data understandable and suitable for analysis. 
```

```python
## Conclusion. 
This dataset is now ready and clean for analysis. 
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
