```python
# Exploratory Data Analysis (EDA)
In this task, I analyzed the dataset to understand patterns and relationships. I used statistical summaries and visualizations like histograms, boxplots, and scatter plots to explore the data.
```

```python
import pandas as pd
df= pd.read_csv("cleaned_data.csv")
df.head()
```

```python
## Summary Statistics 
I used the describe function, to get the summary statiscics of this sentiment dataset, and it displays the Mean, Count, Standard Deviation as well as other important statatistical summary of the numerical values. 
```

```python
df.describe()
```

```python
## Histogram
I used a histogram to visualize the distribution of likes. The values range from around 10 to 80, with most of the data concentrated between 30 and 50. This indicates that the majority of posts receive a moderate level of engagement, while fewer posts having very low or very high likes.
```

```python
import matplotlib.pyplot as plt

df['Likes'].hist()
plt.title("Distribution of Likes")
plt.xlabel("Likes")
plt.ylabel("Frequency")
plt.show()
```

```python
## Boxplots
I used boxplots to show the spread of likes and retweets. Most likes are between 35 and 50. 
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['Likes'])
plt.title("Boxplot of Likes")
plt.ylabel("Likes")
plt.show()
```

```python
This boxplot for retweets indicates that the values are between 18 and 25. There are also some outliers, which means a few posts have very low or very high engagement.
```

```python
plt.boxplot(df['Retweets'])
plt.title("Boxplot of Retweets")
plt.ylabel("Retweets")
plt.show()
```

```python
## Scatter Plot
The scatter plot shows a positive relationship between retweets and likes, meaning that posts with more retweets tend to receive more likes.
```

```python
plt.scatter(df['Retweets'], df['Likes'])
plt.xlabel("Retweets")
plt.ylabel("Likes")
plt.title("Retweets vs Likes")
plt.show()
```

```python
## Correlation
While i performed the correlation analysis, I encountered an error due to non-numeric columns such as text data. I fixed this by selecting only numerical columns for the analysis
The results showed relationships between the likes and retweets variables, which had a strong positive correlation.
```

```python
df_numeric = df.select_dtypes(include=['int64', 'float64'])
df_numeric.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df_numeric.corr(), annot=True)
plt.show()
```

```python
## In conclusion:
Likes are generally higher than retweets, this indicates that users are more likely to like posts than share them.
```
